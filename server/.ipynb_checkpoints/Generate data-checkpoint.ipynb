```python
dt = pd.read_csv('data.csv')
dt = pd.DataFrame(dt, columns=['MCB', 'Ngày công bố', 'Nhóm tuổi', 'G-Related-Cases'])
```

```python
dt
```

```python
end_node = list(dt['MCB'])
group = map(lambda x: x.split('.')[0], list(dt['Nhóm tuổi']))
start_node = list(dt['G-Related-Cases'])
```

```python
graph = {}
for s, t ,g in zip(end_node, start_node, group):
    t = t.replace(" ","")
    graph[s] = [t.split(','), g]
```

```python
len(graph)
```

```python
graph
```

```python
rela = []
for src in graph.keys():
    print(graph[src])
    for trg in graph[src]:
        rela.append([trg[0], trg[1], src])
```

```python
t = pd.DataFrame(rela)
```

```python
t.to_csv('relation.csv')
```
